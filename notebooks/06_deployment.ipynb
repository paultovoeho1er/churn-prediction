```python
import os
import pickle
import numpy as np
import pandas as pd

# Créer les dossiers nécessaires
os.makedirs('app', exist_ok=True)
os.makedirs('app/models', exist_ok=True)
os.makedirs('app/utils', exist_ok=True)

# Copier les modèles dans le dossier app/
import shutil

# Copier le meilleur modèle (XGBoost optimisé par défaut)
if os.path.exists('models/xgboost_best.pkl'):
    shutil.copy('models/xgboost_best.pkl', 'app/models/churn_model.pkl')
    print("✅ Modèle XGBoost copié")
else:
    # Fallback sur Random Forest
    shutil.copy('models/random_forest_best.pkl', 'app/models/churn_model.pkl')
    print("✅ Modèle Random Forest copié")

# Copier le scaler
if os.path.exists('models/scaler.pkl'):
    shutil.copy('models/scaler.pkl', 'app/models/scaler.pkl')
    print("✅ Scaler copié")

# Copier les feature names
if os.path.exists('models/feature_names.pkl'):
    shutil.copy('models/feature_names.pkl', 'app/models/feature_names.pkl')
    print("✅ Feature names copiés")
```
